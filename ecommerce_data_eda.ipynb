```python
import pandas as pd
file=pd.read_csv("ecommerce_customer_churn_dataset.csv")
```

```python
file.columns=file.columns.str.lower()
```

```python
file.isnull().sum()
```

```python
#Replacing Null Values of Age Column To Median
file['age']=file['age'].fillna(file['age'].median())
```

```python
cols=['session_duration_avg','pages_per_session','wishlist_items','returns_rate','product_reviews_written','email_open_rate',
      'discount_usage_rate','social_media_engagement_score','mobile_app_usage','credit_balance','customer_service_calls']
file[cols]=file[cols].fillna(0)
```

```python
file['payment_method_diversity']=file['payment_method_diversity'].fillna(1)
```

```python
max_days=file['days_since_last_purchase'].max()
file['days_since_last_purchase']=file['days_since_last_purchase'].fillna(max_days+1)
```

```python
file['activity_status']=np.where(file['login_frequency']==0,'Inactive','Active')
```

```python
file['purchase_type']=pd.cut(file['total_purchases'],bins=[-1,1,5,10,file['total_purchases'].max()],labels=['One-Time','Low','Medium','High'])
```

```python
file['value_segment']=pd.qcut(file['lifetime_value'],q=3,labels=['Low Value','Medium Value','High Value'])
```

```python
file['recency_bucket']=pd.cut(file['days_since_last_purchase'],bins=[-1,30,90,180,file['days_since_last_purchase'].max()],labels=['30 Days','30-90 Days',
                                                                                                                                  '90-180 Days','180+ Days'])
```
